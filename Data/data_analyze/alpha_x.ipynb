```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd  
import os
from scipy.signal import savgol_filter
import matplotlib.font_manager as font_manager
import matplotlib

def aplha_xing(Ux):
    U_x = []
    for U0 in Ux:
        U_max = max(U0)
        U_average = sum(U0) / len(U0)
        U_x.append(U_max / U_average)
    return U_x

# Set global parameters
matplotlib.rcParams['figure.dpi'] = 100  
plt.rcParams['font.family'] = 'serif'
plt.rcParams['font.serif'] = ['Times New Roman']

# Set math font to Times New Roman
plt.rcParams['mathtext.fontset'] = 'custom'
plt.rcParams['mathtext.rm'] = 'Times New Roman'
plt.rcParams['mathtext.it'] = 'Times New Roman:italic'
plt.rcParams['mathtext.bf'] = 'Times New Roman:bold'

plt.rcParams['font.size'] = 18
plt.rcParams['axes.labelsize'] = 18
plt.rcParams['axes.titlesize'] = 18
plt.rcParams['xtick.labelsize'] = 18
plt.rcParams['ytick.labelsize'] = 18
plt.rcParams['legend.fontsize'] = 18
plt.rcParams['figure.figsize'] = (8, 6)

# Create figure
fig, ax = plt.subplots()

# Color and marker lists
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd', '#8c564b', '#e377c2']
markers = ['o', 's', '^', 'D', 'v', 'p', '*']

# Process 5 groups of data
b_values = [2, 3, 4, 5, 5, 8, 10]
for i, b in enumerate(b_values):
    # Load data file
    data_file = f"data/testH_B{b}.dat"
    data_set = np.float32(np.loadtxt(data_file, unpack=True))
    
    # Process data
    Ux, H = [], []
    for data0 in data_set:
        Ux0 = data0[2:1002]
        Ux.append(Ux0)
        H.append(data0[1])
    
    # Compute U_x
    U_x = aplha_xing(Ux)
    
    # Reorder data
    H = np.array(H)
    U_x = np.array(U_x)
    idx = np.argsort(H)  # Indices for ascending order of H
    H_sorted = H[idx]
    Ux_sorted = U_x[idx]
    
    # Plot data
    ax.scatter(H_sorted, Ux_sorted, color=colors[i], marker=markers[i], s=10, alpha=0.8)
    ax.plot(H_sorted, Ux_sorted, color=colors[i], linestyle='-', linewidth=1.5, label=f'$B$ = {b}(m)', alpha=0.6)       

# Set plot properties
ax.set_xlabel('$H$ (m)')
ax.set_ylabel(r'$\alpha^*$ ($u_{max}/u_{avg}$)')
# ax.set_title('Velocity Ratio Distribution')
# ax.grid(True, linestyle='--', alpha=0.7)
ax.legend(loc='upper center', bbox_to_anchor=(0.5, 1.2), fancybox=True, shadow=True,
          ncol=4, fontsize=14, frameon=False)
# Adjust layout
plt.grid(False)
plt.tight_layout()

os.makedirs('fig', exist_ok=True)

# Save figures to the figsave folder
plt.savefig('fig/fig4.png', dpi=100, bbox_inches='tight')

# Display the plot
plt.show()
```
